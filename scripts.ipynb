```python
import sys
import xml.etree.ElementTree as etree

def get_attr_number(node):
    
    # initialize the attribute counter
    counter = 0
    
    # iterate through node and it's tree
    for i in node.iter():
        
        
        counter += len(i.attrib)

    return counter       
    

if __name__ == '__main__':
    sys.stdin.readline()
    xml = sys.stdin.read()
    tree = etree.ElementTree(etree.fromstring(xml))
    root = tree.getroot()
    print(get_attr_number(root))
```

**2. XML 2**

```python
import xml.etree.ElementTree as etree



maxdepth = 0
def depth(elem, level):
    global maxdepth
    
    # increase the level the current element
    level += 1
    
    # update the global maximum depth if we've gone deeper
    if level > maxdepth:
        maxdepth = level
        
    # loop through all the children of the current element
    for child in elem:
        
        # call the function for each child
        depth(child, level)
        
if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort1' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#

def insertionSort1(n, arr):
    # Write your code here
    
    # save the last value of the array
    last_element = arr[n-1]

    # loop through the from the second last to the initial value of the array
    for i in range(n-2, -1, -1):
        
        # copy the value at the lower index to the current index and print the array
        if arr[i] > last_element:
            arr[i+1] = arr[i]
            
            print(*arr)
            
            
        # indexed value is smaller then insert the last_element at the current index and print the entire array.    
        else:
            arr[i+1] = last_element 
            
            print(*arr)
            
            break
            
    else: # only runs if no 'break' occurred.
        arr[0] = last_element
        print(*arr)
    
if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort1(n, arr)
```
