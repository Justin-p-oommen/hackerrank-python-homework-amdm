**1. Introduction to sets**

```python
def average(array):
    
    a = set(array) #turning array to set a
    
    avrg = sum(a)/len(a) # calculating the average
    
    return f"{avrg:.3f}" # return the average with 3 floating points

if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

**2. Set .discard(), .remove() & .pop()**

```python
n = int(input())
s = set(map(int, input().split()))

N = int(input())    # number of commands

#command = ["pop", "remove", "discard"]

for _ in range(N):

    command = input().split()   # take the commands as input strings

    if command[0] == "pop":
        s.pop()
    elif command[0] == "remove":
        s.remove(int(command[1]))
    elif command[0] == "discard":
        s.discard(int(command[1]))


# print the sum of set s
print(sum(s))
```

**5. The Captain's Room**

```python
# size of each group
K = int(input())

# unordered elements of the room number list
arr = list(map(int, input().split()))


# create a dictionary for counting room -> {room : count}
room_counter = {}


# loop through the list arr
for i in arr:
    
    # if room is repeated in the list, the count updates
    if i in room_counter:  
        room_counter[i] += 1
        
    # initialize all the room count at first
    else:   
        room_counter[i] = 1
        
 
 # loop through the room counter and find which room have count equals 1 and break the loop    
for room, count in room_counter.items():
    if count == 1:
        print(room)
        break
```

**6.  Check Strict Superset**

```python
# Elements of set A
A = set(map(int, input().split()))

# number of other set
n = int(input())

# set the variable to True so that it can continue in the loop if not false
A_superset = True

for _ in range(n):
    N = set(map(int, input().split()))
    
    # check superset condition
    if A > N:
        continue
    else:
        A_superset = False
        break # stop the loop immediately finding the flase condition
        

# print the result that we find an other set which is not a super set
print(A_superset)
```
