```python
from collections import Counter

# number of shoes
X = int(input())

# list of all the shoe sizes in the shop.
shoe_size = list(map(int, input().split()))

# Counter - A dictionary contains the size and the number of shoes of that size
count_shoes = Counter(shoe_size)

# initialize the money earned
money_earned = 0

# Number of customers
N = int(input())


# Loop through the customer
for _ in range(N):
    
    # take the size desired by customer and the price
    size, price = map(int, input().split())
    
    # check the size desired by the customer is ok. if yes then add to the money
    if count_shoes[size] > 0:
        
        money_earned += price
        
        # decrease the number of shoes from the inventory
        count_shoes[size] -= 1    
    
    
# Total sum of money earned
print(money_earned)
    
    
```

**2. DefaultDict Tutorial**

```python
from collections import defaultdict

n, m = map(int, input().split())

# Group A
A = [input() for _ in range(n)]

# Group B
B = [input() for _ in range(m)]


# Create the dictionary (Mapping index of A)
dd_A = defaultdict(list)


for index, word in enumerate(A, start=1):
    dd_A[word].append(index)

# Loop through B
for word in B:
    
    if word in dd_A:
        
        # unpacking the list of ind
        print(*dd_A[word]) 
        
    else:
        print(-1)
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'birthdayCakeCandles' function below.
#
# The function is expected to return an INTEGER.
# The function accepts INTEGER_ARRAY candles as parameter.
#

def birthdayCakeCandles(candles):
    # Write your code here
    
    # initialize the candle counter
    candle_counter = 0
    
    # find the maximum number in the list
    max_candle = max(candles)
    
    # loop through the list
    for i in candles:
        
        # check the list element to the maximum number and if yes, update the counter
        if i == max_candle:
            candle_counter += 1
            
    # return the total number of tallest candle        
    return candle_counter
    

if __name__ == '__main__':
   

    candles_count = int(input().strip())

    candles = list(map(int, input().rstrip().split()))

    result = birthdayCakeCandles(candles)

    print(result)
```

## insertionsort1

```python
#!/bin/python3

import math
import os
import random
import re
import sys

#
# Complete the 'insertionSort1' function below.
#
# The function accepts following parameters:
#  1. INTEGER n
#  2. INTEGER_ARRAY arr
#

def insertionSort1(n, arr):
    # Write your code here
    
    # save the last value of the array
    last_element = arr[n-1]

    # loop through the from the second last to the initial value of the array
    for i in range(n-2, -1, -1):
        
        # copy the value at the lower index to the current index and print the array
        if arr[i] > last_element:
            arr[i+1] = arr[i]
            
            print(*arr)
            
            
        # indexed value is smaller then insert the last_element at the current index and print the entire array.    
        else:
            arr[i+1] = last_element 
            
            print(*arr)
            
            break
            
    else: # only runs if no 'break' occurred.
        arr[0] = last_element
        print(*arr)
    
if __name__ == '__main__':
    n = int(input().strip())

    arr = list(map(int, input().rstrip().split()))

    insertionSort1(n, arr)
```
